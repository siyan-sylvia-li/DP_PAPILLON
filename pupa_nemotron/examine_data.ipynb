```python
all_pii_counts = []
for entry in curr_data:
    all_piis = set(entry["pii_units"].lower().split("||")).union(set(entry["pii_gliner"].lower().split("||")))
    all_pii_counts.append(len(all_piis))

from collections import Counter
Counter(all_pii_counts)
```

```python
for entry in curr_data:
    if len(set(entry["pii_units"].lower().split("||")).union(set(entry["pii_gliner"].lower().split("||")))) == 34:
        print(entry)
```
